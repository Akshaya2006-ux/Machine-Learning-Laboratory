```python
import numpy as np

# Number of samples
n = 10000

# Store samples
samples = np.zeros((n, 2))

# Initial values
x = 0
y = 0

# Gibbs Sampling
for i in range(n):

    # Sample X given Y
    x = np.random.normal(y, 1)

    # Sample Y given X
    y = np.random.normal(x, 1)

    samples[i] = [x, y]


# Mean of generated samples
print("Mean of X:", np.mean(samples[:, 0]))
print("Mean of Y:", np.mean(samples[:, 1]))

# Standard deviation
print("Standard deviation of X:", np.std(samples[:, 0]))
print("Standard deviation of Y:", np.std(samples[:, 1]))

# First 10 samples
print("\nFirst 10 samples:")
print(samples[:10])
```
