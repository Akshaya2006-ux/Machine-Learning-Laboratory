```python
import numpy as np


X = np.array([
    ["Sunny", "Hot", "High"],
    ["Sunny", "Hot", "High"],
    ["Overcast", "Hot", "High"],
    ["Rain", "Mild", "High"],
    ["Rain", "Cool", "Normal"],
    ["Rain", "Cool", "Normal"],
    ["Overcast", "Cool", "Normal"],
    ["Sunny", "Mild", "High"],
    ["Sunny", "Cool", "Normal"],
    ["Rain", "Mild", "Normal"]
])

y = np.array([
    "No", "No", "Yes", "Yes", "Yes",
    "No", "Yes", "No", "Yes", "Yes"
])

def calculate_prior(y):
    classes, counts = np.unique(y, return_counts=True)

    prior = {}

    for c, count in zip(classes, counts):
        prior[c] = count / len(y)

    return prior


prior = calculate_prior(y)

def calculate_probability(X, y):
    probability = {}

    classes = np.unique(y)

    for c in classes:

        probability[c] = {}

        X_class = X[y == c]

        for feature in range(X.shape[1]):

            probability[c][feature] = {}

            values = np.unique(X[:, feature])

            for value in values:

                count = np.sum(X_class[:, feature] == value)

                probability[c][feature][value] = (
                    count / len(X_class)
                )

    return probability


probability = calculate_probability(X, y)


def predict(sample):

    classes = np.unique(y)

    posterior = {}

    for c in classes:

        prob = prior[c]

    
        for feature in range(len(sample)):

            value = sample[feature]

            if value in probability[c][feature]:

                prob *= probability[c][feature][value]

            else:
                prob = 0

        posterior[c] = prob

    return max(posterior, key=posterior.get)


sample = ["Sunny", "Cool", "Normal"]

prediction = predict(sample)

print("Test Sample:", sample)
print("Prediction:", prediction)
```
