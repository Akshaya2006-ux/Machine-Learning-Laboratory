```python
states = ('good', 'moderate', 'fail')

initial = {
    'good': 0.5,
    'moderate': 0.25,
    'fail': 0.25
}

emission = {
    'good': {'happy': 0.4, 'worry': 0.1, 'blink': 0.2, 'fine': 0.3},
    'moderate': {'happy': 0.3, 'worry': 0.2, 'blink': 0.3, 'fine': 0.2},
    'fail': {'happy': 0.1, 'worry': 0.4, 'blink': 0.3, 'fine': 0.2}
}

def naive_bayes(obs):
    prob = {}

    for s in states:
        p = initial[s]

        for o in obs:
            p *= emission[s][o]

        prob[s] = p

    best_state = max(prob, key=prob.get)

    return best_state, prob


# Example
observations = ['happy', 'blink', 'worry', 'fine']

state, probabilities = naive_bayes(observations)

print("Probabilities:", probabilities)
print("Predicted state:", state)
```
