```python
states = ('good', 'moderate', 'fail')

initial = {
    'good': 0.5,
    'moderate': 0.25,
    'fail': 0.25
}

transition = {
    'good': {'good': 0.4, 'moderate': 0.3, 'fail': 0.3},
    'moderate': {'moderate': 0.25, 'good': 0.6, 'fail': 0.15},
    'fail': {'good': 0.3, 'moderate': 0.4, 'fail': 0.3}
}

emission = {
    'good': {'happy': 0.4, 'worry': 0.1, 'blink': 0.2, 'fine': 0.3},
    'moderate': {'happy': 0.3, 'worry': 0.2, 'blink': 0.3, 'fine': 0.2},
    'fail': {'happy': 0.1, 'worry': 0.4, 'blink': 0.3, 'fine': 0.2}
}


def viterbi(obs):
    # Initialize probabilities
    prob = {
        s: initial[s] * emission[s][obs[0]]
        for s in states
    }

    paths = {s: [s] for s in states}

    # Process remaining observations
    for o in obs[1:]:
        new_prob = {}
        new_paths = {}

        for s in states:
            p, prev = max(
                (prob[x] * transition[x][s] * emission[s][o], x)
                for x in states
            )

            new_prob[s] = p
            new_paths[s] = paths[prev] + [s]

        prob = new_prob
        paths = new_paths

    # Return the most likely state sequence
    best_state = max(prob, key=prob.get)

    return paths[best_state], prob[best_state]


# Example
observations = ['happy', 'blink', 'worry', 'fine']

best_path, best_probability = viterbi(observations)

print("Best state sequence:", best_path)
print("Probability:", best_probability)
```
