```python
from sklearn.metrics import confusion_matrix, accuracy_score, precision_score, recall_score, f1_score

y_true = ["Spam", "Spam", "Not Spam", "Not Spam", "Spam", "Not Spam", "Spam", "Not Spam"]


y_pred = ["Spam", "Not Spam", "Spam", "Not Spam", "Spam", "Not Spam", "Spam", "Not Spam"]

cm = confusion_matrix(y_true, y_pred, labels=["Spam", "Not Spam"])

print("Confusion Matrix:")
print(cm)


TP = cm[0][0]
FN = cm[0][1]
FP = cm[1][0]
TN = cm[1][1]

print("\nTP =", TP)
print("TN =", TN)
print("FP =", FP)
print("FN =", FN)

print("\nAccuracy :", accuracy_score(y_true, y_pred))
print("Precision:", precision_score(y_true, y_pred, pos_label="Spam"))
print("Recall   :", recall_score(y_true, y_pred, pos_label="Spam"))
print("F1 Score :", f1_score(y_true, y_pred, pos_label="Spam"))
```
