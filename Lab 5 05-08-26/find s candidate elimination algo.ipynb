```python


data = [
    ["sky", "airtemp", "humidity", "wind", "water", "forecast", "enjoysport"],
    ["sunny", "warm", "normal", "strong", "warm", "same", "yes"],
    ["sunny", "warm", "high", "strong", "warm", "same", "yes"],
    ["rainy", "cold", "high", "strong", "warm", "change", "no"],
    ["sunny", "warm", "high", "strong", "cool", "change", "yes"],
]

concepts = [row[:-1] for row in data[1:]]
target = [row[-1] for row in data[1:]]

num_attr = len(concepts[0])


S = ['Ø'] * num_attr
G = [['?'] * num_attr]


def consistent(h, x):
    return all(h[i] == '?' or h[i] == x[i] for i in range(num_attr))


for i, example in enumerate(concepts):


    if target[i].lower() == "yes":

        if S[0] == 'Ø':
            S = example.copy()
        else:
            for j in range(num_attr):
                if S[j] != example[j]:
                    S[j] = '?'

        G = [g for g in G if consistent(g, example)]

    else:

        new_G = []

        for g in G:

            if consistent(g, example):

                for j in range(num_attr):

                    if g[j] == '?' and S[j] != '?' and S[j] != example[j]:
                        h = g.copy()
                        h[j] = S[j]
                        new_G.append(h)

            else:
                new_G.append(g)

   
        G = []
        for h in new_G:
            if h not in G:
                G.append(h)


print("Specific Hypothesis:")
print(S)

print("\nGeneral Hypotheses:")
for g in G:
    print(g)
```
